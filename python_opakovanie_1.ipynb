## 1. Premenné a priradenie

Premenná je názov, ktorý odkazuje na objekt v pamäti. V Pythone sú hodnoty reprezentované objektmi.

V Pythone sa typ premennej nedeklaruje vopred; určuje ho priradená hodnota.

````python
# Celé číslo (int)
x = 5
print(x)

# Desatinné číslo (float)
y = 3.14
print(y)

# Záporné číslo
z = -10
print(z)
````

Funkcia `type()` vracia typ hodnoty, na ktorú premenná odkazuje:

````python
x = 5
print(type(x))   # int = celé číslo

y = 3.14
print(type(y))   # float = desatinné číslo
````

### 1.1 Zmena hodnoty premennej

Hodnotu premennej možno kedykoľvek nahradiť inou hodnotou. Premenná následne odkazuje na novú hodnotu.

````python
x = 10
print("x je", x)

x = 20            # prepíšeme starú hodnotu
print("x je teraz", x)
````

Premennú možno aktualizovať aj na základe jej aktuálnej hodnoty:

````python
x = 20
x = x + 5         # vezmi aktuálne x (20), pričítaj 5, ulož späť
print("x + 5 =", x)

y = 20
y = y - 8
print("y - 8 =", y)

z = 4
z = z * 2
print("z * 2 =", z)

a = 3
a = a ** 2         # ** znamená umocnenie
print("a ** 2 =", a)
````

### 1.2 Skrátený zápis operácií

Zápis `x = x + 5` možno skrátiť na `x += 5`. Tento princíp platí pre všetky aritmetické operácie:

| Dlhý zápis | Skrátený zápis |
|---|---|
| `x = x + 3` | `x += 3` |
| `x = x - 3` | `x -= 3` |
| `x = x * 5` | `x *= 5` |
| `x = x / 5` | `x /= 5` |
| `x = x ** 2` | `x **= 2` |
| `x = x % 2` | `x %= 2` |

Python nepozná operátory `x++` a `x--` známe z iných jazykov. Používajú sa zápisy `x += 1` a `x -= 1`.

````python
x = 10
x += 3
print("po += 3:", x)

x -= 5
print("po -= 5:", x)

x *= 4
print("po *= 4:", x)

x /= 2
print("po /= 2:", x)
````

## 2. Výstup pomocou `print()`

Funkcia `print()` slúži na výpis hodnôt.

### 2.1 Výpis viacerých hodnôt

Do `print()` možno zapísať viacero hodnôt oddelených čiarkami. Hodnoty sa vypíšú oddelené medzerami:

````python
a = 1
b = 2
print(a, b)

x = 5
y = 10
print("x =", x, "a y =", y)
````

### 2.2 Formátovaný výstup – f-reťazce

Na presnejšiu kontrolu výstupu slúžia f-reťazce. Pred úvodzovky sa zapisuje písmeno `f` a premenná alebo výraz sa uzavrie do `{}`:

````python
meno = "Jano"
vek = 16
print(f"Volám sa {meno} a mám {vek} rokov.")
````

````python
# Do {} možno vložiť aj výraz
x = 15
y = 20
print(f"Súčet x + y = {x + y}")
print(f"Rozdiel x - y = {x - y}")
````

#### Zaokrúhľovanie v f-reťazcoch

Ak sa má desatinné číslo zobraziť na určitý počet miest, používa sa formát `:.Xf`, kde `X` je počet desatinných miest:

````python
pi = 3.141592653589793

print(f"Pi na 2 miesta: {pi:.2f}")
print(f"Pi na 4 miesta: {pi:.4f}")
print(f"Pi na 0 miest:  {pi:.0f}")
````

## 3. Základné dátové typy

### 3.1 Celé čísla (`int`) a desatinné čísla (`float`)

| Operácia | Zápis | Príklad | Výsledok |
|---|---|---|---|
| Sčítanie | `+` | `3 + 2` | `5` |
| Odčítanie | `-` | `3 - 2` | `1` |
| Násobenie | `*` | `3 * 2` | `6` |
| Delenie | `/` | `7 / 2` | `3.5` |
| Celočíselné delenie | `//` | `7 // 2` | `3` |
| Modulo (zvyšok) | `%` | `7 % 2` | `1` |
| Umocnenie | `**` | `3 ** 2` | `9` |

````python
# Porovnanie operácií
print("7 / 2  =", 7 / 2)    # bežné delenie → vždy float
print("7 // 2 =", 7 // 2)   # celočíselné delenie → zahodí zvyšok
print("7 % 2  =", 7 % 2)    # modulo → vráti len zvyšok
````

### 3.2 Boolovské hodnoty (`bool`)

Boolovské (logické) premenné majú iba dve hodnoty: `True` (pravda) a `False` (nepravda). Vznikajú napríklad pri porovnávaní:

````python
print(5 > 3)     # True
print(5 == 3)    # False
print(5 != 3)    # True  (nie je rovné)
````

````python
# Logické operácie
t = True
f = False

print("t and f =", t and f)     # True, len ak sú OBE pravdivé
print("t or f  =", t or f)      # True, ak je ASPOŇ JEDNA pravdivá
print("not t   =", not t)       # opak (negácia)
````

### 3.3 Reťazce (`str`)

Reťazce sú texty uzavreté v úvodzovkách – jednoduchých `'...'` alebo dvojitých `"..."`.

````python
meno = "Katka"
pozdrav = 'Ahoj'

print(type(meno))         # <class 'str'>
print(len(meno))          # dĺžka reťazca: 5
````

````python
# Spájanie reťazcov operátorom +
meno = "Katka"
pozdrav = "Ahoj"

veta = pozdrav + ", " + meno + "!"
print(veta)
````

````python
# Užitočné funkcie na reťazcoch
text = "ahoj svet  "

print(text.strip())        # odstráni medzery na krajoch
print(text.upper())        # všetko veľkými písmenami
print(text.capitalize())   # prvý znak veľkým
print(text.replace("svet", "Python"))  # nahradí slovo
````

## 4. Podmienky: `if`, `elif`, `else`

Podmienky umožňujú vykonať rôzne časti kódu podľa výsledku vyhodnotenia logického výrazu.

```
if podmienka:
    # tento blok sa vykoná, ak je podmienka True
elif iná_podmienka:
    # tento blok sa vykoná, ak je iná_podmienka True
else:
    # tento blok sa vykoná, ak žiadna podmienka nebola True
```

Za podmienkou musí byť dvojbodka `:` a telo bloku sa odsadí štyrmi medzerami. Odsadenie nie je len pre prehľadnosť, je súčasťou syntaxe Pythonu.

````python
vek = 17

if vek >= 18:
    print("Si plnoletý.")
else:
    print("Ešte nie si plnoletý.")
````

````python
# Viac vetiev s elif
znamka = 2

if znamka == 1:
    print("Výborne!")
elif znamka == 2:
    print("Chválitebne.")
elif znamka == 3:
    print("Dobre.")
elif znamka == 4:
    print("Dostatočne.")
else:
    print("Nedostatočne!")
````

````python
# Kombinovanie podmienok pomocou and / or
x = 1
a = 7

if x == 1 and a < 10:
    print("x je 1 a zároveň a je menšie ako 10")
elif a > 17 or x == 0:
    print("a je väčšie ako 17 alebo x je 0")
else:
    print("Ani jedna podmienka nebola splnená")
````

# Rozšírenie – funkcie
## 7.1 Význam funkcií

Pri opakovanom výpočte obvodu obdĺžnika pre rôzne rozmery je opakovanie toho istého vzorca neefektívne. Funkcia umožňuje vzorec definovať raz a následne ho volať s rôznymi argumentmi.




````python
# BEZ funkcie – opakujeme sa:
print(2 * (5 + 3))
print(2 * (10 + 4))
print(2 * (7 + 2))

# S funkciou – vzorec je definovaný raz a funkcia sa volá viackrát:
def obvod(a, b):
    return 2 * (a + b)

print(obvod(5, 3))
print(obvod(10, 4))
print(obvod(7, 2))
````


---

## 7.2 Anatómia funkcie



````python
def nazov_funkcie(parameter1, parameter2):
    # telo funkcie (odsadené!)
    vysledok = parameter1 + parameter2
    return vysledok
````


- `def` – kľúčové slovo na definovanie funkcie
- `nazov_funkcie` – názov, ktorým sa funkcia volá
- `(parameter1, parameter2)` – vstupné hodnoty; počet parametrov môže byť 0, 1, 2, ...
- `:` – dvojbodka na konci riadku definície
- telo je odsadené štyrmi medzerami
- `return` – vráti výsledok do miesta volania funkcie

## 7.6 Rozdiel medzi `return` a `print`

Rozdiel je dôležitý najmä pri ďalšom spracovaní výsledku:


````python
# Funkcia s return – vráti hodnotu na ďalšie spracovanie
def scitaj(a, b):
    return a + b

x = scitaj(3, 4)       # x je teraz 7
print(x * 2)           # 14 – výsledok je možné spracovať

# Funkcia s print – hodnotu vypíše, ale nevráti ju
def scitaj_print(a, b):
    print(a + b)

y = scitaj_print(3, 4)  # vypíše 7
print(y)                 # None – funkcia nevrátila hodnotu
````

Ak sa má výsledok ďalej spracovávať, funkcia ho vracia príkazom `return`. Na výpis hodnoty slúži `print`.
