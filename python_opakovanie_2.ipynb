## 2. Funkcia `range()` – generátor čísel

`range()` vytvára postupnosť celých čísel. Samotný objekt hodnoty nezobrazuje; jeho obsah možno zobraziť napríklad konverziou na zoznam.

Na zobrazenie hodnôt sa objekt `range()` prevedie na zoznam funkciou `list()`.

### 2.1 `range(koniec)` – jeden argument

Generuje čísla od 0 po `koniec`, pričom koniec nie je zahrnutý.

````python
print(list(range(5)))    # 0, 1, 2, 3, 4  (NIE 5!)
print(list(range(10)))   # 0, 1, 2, ..., 9
````

`range(5)` vytvorí päť čísel – 0, 1, 2, 3 a 4. Piatka už v nich nie je, preto horná hranica končí o jedno skôr.

### 2.2 `range(začiatok, koniec)` – dva argumenty

Ak postupnosť nemá začínať nulou, zadáva sa aj počiatočná hodnota.

````python
print(list(range(1, 6)))    # 1, 2, 3, 4, 5
print(list(range(2, 10)))   # 2, 3, 4, 5, 6, 7, 8, 9
print(list(range(10, 15)))  # 10, 11, 12, 13, 14
````

### 2.3 `range(začiatok, koniec, krok)` – tri argumenty

Tretí argument určuje veľkosť kroku. Bez explicitného zadania má krok hodnotu 1.

````python
print(list(range(0, 10, 2)))   # párne: 0, 2, 4, 6, 8
print(list(range(1, 10, 2)))   # nepárne: 1, 3, 5, 7, 9
print(list(range(0, 50, 10)))  # desiatky: 0, 10, 20, 30, 40
print(list(range(2, 10, 3)))   # krok 3: 2, 5, 8
````

### 2.4 `range` so záporným krokom

Pri odpočítavaní musí byť krok záporný a počiatočná hodnota väčšia ako koncová hodnota.

````python
print(list(range(10, 0, -1)))   # 10, 9, 8, ..., 1
print(list(range(5, -1, -1)))   # 5, 4, 3, 2, 1, 0
print(list(range(20, 0, -5)))   # 20, 15, 10, 5
````

### Prehľad `range()`

| Zápis | Výsledok | Vysvetlenie |
|---|---|---|
| `range(5)` | 0, 1, 2, 3, 4 | od 0, krok 1 |
| `range(2, 7)` | 2, 3, 4, 5, 6 | od 2 po 7, krok 1 |
| `range(0, 10, 2)` | 0, 2, 4, 6, 8 | od 0 po 10, krok 2 |
| `range(10, 0, -1)` | 10, 9, ..., 1 | odpočítavanie |

## 3. Cyklus `for`

Cyklus `for` vykoná blok kódu pre každý prvok v nejakej postupnosti.

```
for premenna in postupnost:
    # telo cyklu (odsadené!)
    # vykoná sa raz pre každý prvok
```

Syntax vyžaduje:
- dvojbodku `:` za postupnosťou
- telo cyklu odsadené štyrmi medzerami, rovnako ako pri podmienkach

### 3.1 Najjednoduchší `for` cyklus

````python
# Vypíše čísla 0 až 4
for i in range(5):
    print(i)
````

Postup vykonávania:
1. `range(5)` vytvorí postupnosť: 0, 1, 2, 3, 4
2. Prvý priebeh: `i = 0`, vykoná sa `print(0)`
3. Druhý priebeh: `i = 1`, vykoná sa `print(1)`
4. ... a tak ďalej, až kým sa nevyčerpajú všetky čísla.

### 3.2 `for` s rôznymi variantmi `range()`

````python
# Čísla od 1 do 5
print("Čísla 1 až 5:")
for i in range(1, 6):
    print(i)
````

````python
# Párne čísla od 2 do 10
print("Párne čísla:")
for i in range(2, 11, 2):
    print(i)
````

````python
# Odpočítavanie
print("Odpočítavanie:")
for i in range(5, 0, -1):
    print(i)
print("Štart!")
````

### 3.3 Obsah tela cyklu

V tele cyklu možno použiť výpočty, podmienky aj volania funkcií.

````python
# Výpočet druhých mocnín
for i in range(1, 6):
    print(f"{i} na druhú je {i ** 2}")
````

````python
# Podmienka vnútri cyklu – ktoré čísla sú párne a ktoré nepárne?
for i in range(1, 11):
    if i % 2 == 0:
        print(f"{i} je párne")
    else:
        print(f"{i} je nepárne")
````

### 3.4 Sčítavanie v cykle – akumulátor

Na súčet viacerých čísel sa používa premenná – akumulátor – ktorá sa v každom priechode cyklu zvyšuje.

````python
# Súčet čísel od 1 do 10
sucet = 0                     # akumulátor – začíname na nule

for i in range(1, 11):
    sucet += i                # v každom priechode pripočítame i
    print(f"Priechod {i}: sucet = {sucet}")

print(f"\nVýsledný súčet 1 + 2 + ... + 10 = {sucet}")
````

### 3.5 `for` cez reťazec

Cyklus `for` neprechádza len číslami. Môže prejsť aj napríklad cez písmená v reťazci:

````python
slovo = "Python"

for pismeno in slovo:
    print(pismeno)
````

## 4. Cyklus `while`

Cyklus `while` opakuje blok kódu, kým je podmienka pravdivá (`True`).

```
while podmienka:
    # telo cyklu (odsadené!)
    # vykonáva sa, kým je podmienka True
```

### Kedy `while` a kedy `for`?

| Situácia | Použite |
|---|---|
| Počet opakovaní je známy | `for` |
| Je potrebné prejsť postupnosťou hodnôt | `for` |
| Počet opakovaní určuje podmienka | `while` |
| Opakovanie má platiť, kým platí podmienka | `while` |

### 4.1 Najjednoduchší `while` cyklus

````python
# Vypíše čísla 0 až 4
i = 0

while i < 5:
    print(i)
    i += 1       # bez tohto riadka by cyklus bežal donekonečna
````

Postup vykonávania:
1. `i = 0` – inicializácia riadiacej premennej
2. Je `i < 5`? Áno (0 < 5) → vykoná sa telo: vypíše 0, zvýši `i` na 1
3. Je `i < 5`? Áno (1 < 5) → vykoná sa telo: vypíše 1, zvýši `i` na 2
4. ... pokračuje ...
5. Je `i < 5`? Nie (5 < 5 je False), takže cyklus končí

Ak sa riadiaca premenná v tele cyklu nezvyšuje, cyklus je nekonečný. Jeho vykonávanie je potrebné zastaviť príslušným tlačidlom vedľa bunky.

### 4.2 `while` – počet opakovaní nie je známy

Tu je príklad, ktorý sa jednoducho nenahrádza `for` cyklom:

````python
# Opakované umocňovanie: začíname na 2, umocňujeme na druhú,
# kým neprekročíme 100
c = 2

while c < 100:
    c **= 2
    print(c)
````

Počet priebehov nebol na začiatku známy; určuje ho podmienka ukončenia cyklu.

````python
# Koľkokrát musíme vydeliť číslo dvoma, kým dostaneme menej ako 1?
cislo = 1000
pocet = 0

while cislo >= 1:
    cislo /= 2
    pocet += 1

print(f"Delili sme {pocet}-krát, zostalo {cislo}")
````

### 4.3 `while` s vstupom od používateľa

Cyklus `while` sa používa aj pri opakovanom čítaní vstupu do zadania terminálnej hodnoty.

````python
# Hodnota 0 ukončuje zadávanie čísel.
sucet = 0
cislo = int(input("Zadajte číslo (0 = koniec): "))

while cislo != 0:
    sucet += cislo
    cislo = int(input("Zadajte číslo (0 = koniec): "))

print(f"Súčet zadaných čísel je {sucet}")
````

## 5. Porovnanie `for` a `while`

Rovnaký výstup možno často dosiahnuť oboma cyklami:

````python
# Výpis čísel 0 až 4 pomocou FOR
print("=== FOR ===")
for i in range(5):
    print(i)

# Ten istý výstup pomocou WHILE
print("=== WHILE ===")
i = 0
while i < 5:
    print(i)
    i += 1
````

Verzia s `for` je kratšia, pretože riadiaca premenná sa nemusí inicializovať ani zvyšovať ručne. Pri známom počte opakovaní je preto zvyčajne vhodnejšia.

## 6. `break` a `continue` – riadenie cyklu

Niekedy treba cyklus predčasne ukončiť alebo preskočiť jeden priechod.

### 6.1 `break` – okamžité ukončenie cyklu

````python
# Hľadáme prvé číslo deliteľné 7 v rozsahu 1-100
for i in range(1, 101):
    if i % 7 == 0:
        print(f"Prvé číslo deliteľné 7 je {i}")
        break    # našli sme, končíme cyklus
````

### 6.2 `continue` – preskoč zvyšok a pokračuj ďalším priebehom

````python
# Vypíš čísla 1-10, ale preskoč číslo 5
for i in range(1, 11):
    if i == 5:
        continue    # preskočí print a ide na ďalšie i
    print(i)
````

## 7. Cykly a funkcie – kombinácia

Cykly možno použiť v tele funkcie; funkcie možno volať aj z tela cyklu.

````python
def sucet_od_do(od, do):
    """Spočíta súčet celých čísel od 'od' po 'do' (vrátane)."""
    vysledok = 0
    for i in range(od, do + 1):
        vysledok += i
    return vysledok

print(sucet_od_do(1, 10))    # 55
print(sucet_od_do(1, 100))   # 5050
````

````python
def je_prvocislo(n):
    """Zistí, či je číslo n prvočíslo."""
    if n < 2:
        return False
    for i in range(2, n):
        if n % i == 0:
            return False     # našli sme deliteľa → nie je prvočíslo
    return True              # nenašli sme deliteľa → je prvočíslo

# Test
for cislo in range(1, 21):
    if je_prvocislo(cislo):
        print(f"{cislo} je prvočíslo")
````
